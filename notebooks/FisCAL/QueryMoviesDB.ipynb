# Query moviesDB

Read up to 100 rows from `dbo.moviesDB` in the `sqldedpool1` dedicated SQL pool using the Azure Synapse Spark connector.

```python
from notebookutils import mssparkutils

jdbc_url = (
    "jdbc:sqlserver://synw-demoworkspace-001.sql.azuresynapse.net:1433;"
    "database=sqldedpool1;"
    "encrypt=true;"
    "trustServerCertificate=false;"
    "hostNameInCertificate=*.sql.azuresynapse.net;"
    "loginTimeout=30;"
)

query = """
SELECT TOP (100)
    [movie],
    [title],
    [genres],
    [year],
    [Rating],
    [Rotton Tomato] AS [Rotton_Tomato]
FROM [dbo].[moviesDB]
"""

movies_df = (
    spark.read
    .format("jdbc")
    .option("url", jdbc_url)
    .option("query", query)
    .option("accessToken", mssparkutils.credentials.getToken("DW"))
    .load()
)

display(movies_df)
```
