# Query moviesDB

Read up to 100 rows from a dedicated SQL pool table using JDBC and Microsoft Entra authentication. Update the configuration values before running the notebook.

```python
from notebookutils import mssparkutils

workspace_name = "<your-synapse-workspace-name>"
dedicated_pool = "<your-dedicated-sql-pool-name>"
schema_name = "dbo"
table_name = "moviesDB"
sql_port = 1433

jdbc_url = (
    f"jdbc:sqlserver://{workspace_name}.sql.azuresynapse.net:{sql_port};"
    f"database={dedicated_pool};"
    "encrypt=true;"
    "trustServerCertificate=false;"
    "hostNameInCertificate=*.sql.azuresynapse.net;"
    "loginTimeout=30;"
)

query = f"""
SELECT TOP (100)
    [movie],
    [title],
    [genres],
    [year],
    [Rating],
    [Rotten_Tomato]
FROM [{schema_name}].[{table_name}]
"""

movies_df = (
    spark.read
    .format("jdbc")
    .option("url", jdbc_url)
    .option("query", query)
    .option("accessToken", mssparkutils.credentials.getToken("DW"))
    .load()
)

display(movies_df)
```
