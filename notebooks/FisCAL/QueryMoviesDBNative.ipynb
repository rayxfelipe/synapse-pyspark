# Query moviesDB with the native Synapse connector

Read up to 100 rows from `dbo.moviesDB` in the `sqldedpool1` dedicated SQL pool using the native Azure Synapse Dedicated SQL Pool Connector for Apache Spark.

```python
import com.microsoft.spark.sqlanalytics
from com.microsoft.spark.sqlanalytics.Constants import Constants

movies_df = (
    spark.read
    .option(Constants.SERVER, "synw-demoworkspace-001.sql.azuresynapse.net")
    .synapsesql("sqldedpool1.dbo.moviesDB")
)

display(movies_df.limit(100))
```
